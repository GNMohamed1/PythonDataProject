```python
import pandas as pd
```

```python
air_quality_df = pd.read_csv("../data/airquality.csv")
air_quality_df.head()
```

```python
green_covarage_df = pd.read_csv("../data/greenCovarage.csv")
green_covarage_df.head()
```

```python
green_covarage_df = green_covarage_df[['id', 'green_coverage_pct']]
green_covarage_df.head()
```

```python
merged_df = pd.merge(left=air_quality_df, right=green_covarage_df, on='id', how="left")
merged_df.head()
```

```python
districts_df = pd.read_csv("../data/districts.csv")
districts_df.head()
```

```python
locations_df = pd.read_csv("../data/locations.csv")
locations_df.head()
```

```python
merged_df = pd.merge(left=merged_df, right=locations_df, how="left", on="id")
merged_df.head()
```

```python
merged_df = pd.merge(left=merged_df, right=districts_df, how="left", on="lat")
```

```python
merged_df.head()
```

```python
final_df = merged_df[['name_en', 'name_ar', 'pm10', 'pm2_5', 'us_aqi', 'us_pm2_5', 'us_pm10', 'us_no2', 'green_coverage_pct', 'gov_en', 'gov_ar']]
```

```python
final_df.head()
```

```python
final_df.to_csv('../data/qol.csv', sep=",", index=False, header=True)
```
